# Problem 15 - Traveling Salesperson Problem (TSP)

## Introduction

The Traveling Salesperson Problem (TSP) is about finding the shortest possible route between a group of cities. We need to visit every city exactly once and return to the starting city.

In this project, we will test different algorithms to find the shortest route and compare their speed and results.

First, we will create a distance matrix and a function to calculate the total distance of a route.

## Distance Matrix
The matrix shows the distance between every two cities. The row is the city we travel from, and the column is the city we travel to. For example, D[1][3] = 25 means City 1 to City 3 has a distance of 25.

```python

D = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

print("Number of cities:", len(D))
```


## Calculating the Total Route Distance

route_cost() adds the distances between consecutive cities in a route.

For example, we have this route:

**0 → 1 → 3 → 2 → 0**

Distance is:

**10 + 25 + 30 + 15 = 80**

```python
def route_cost(route, D):
    total = 0

    for i in range(len(route) - 1):
        from_city = route[i]
        to_city = route[i + 1]

        total += D[from_city][to_city]

    return total

route = [0, 1, 3, 2, 0]

print("Route:", route)
print("Total distance:", route_cost(route, D))
```


## Stage 1 - Brute Force Algorithm

Brute Force is the first method we use to solve TSP.

The idea is simple. We generate all possible routes, calculate the total distance of each route, and choose the shortest one.

Since our distance matrix is symmetric, going through a route in reverse gives the same distance. So we don't need to check both directions.

**Advantages:**
- Always finds the optimal route.

**Disadvantages:**
- Gets very slow with more cities. 
- Time complexity grows factorially: O(n!).


```python

from itertools import permutations


def brute_force(D, show_routes=False):
    n = len(D)
    best_distance = float("inf")
    best_route = []

    # Generate all possible orders of cities except city 0
    for order in permutations(range(1, n)):

        # Skip reversed routes because distances are symmetric
        if len(order) > 1 and order[0] > order[-1]:
            continue

        # Start and finish at city 0
        route = [0] + list(order) + [0]

        # Calculate total distance
        distance = route_cost(route, D)

        if show_routes:
            print("Route:", route, "Distance:", distance)

        # Save the shortest route
        if distance < best_distance:
            best_distance = distance
            best_route = route

    return best_route, best_distance


best_route, best_distance = brute_force(D, show_routes=True)

print("\nBest route:", best_route)
print("Shortest distance:", best_distance)
```


## Stage 2 - Nearest Neighbor Algorithm

Nearest Neighbor is another method to solve the Traveling Salesperson Problem (TSP).

Unlike Brute Force, which checks all possible routes, Nearest Neighbor always chooses the closest city that we haven't visited yet.

**How it works:**
1. Start from a city.
2. Find the closest city that hasn't been visited.
3. Move to that city.
4. Repeat until all cities are visited.
5. Return to the starting city.

This algorithm is much faster than Brute Force because it doesn't check every possible route.

**Advantages:**
- Fast and easy to implement.
- Works well with a large number of cities.
- Time complexity: O(n²).

**Disadvantages:**
- Doesn't always find the shortest possible route.

```python
def nearest_neighbor(D, start=0):
    n = len(D)
    route = [start]
    unvisited = list(range(n))
    unvisited.remove(start)
    current = start


    while unvisited:

        # Find the closest unvisited city
        next_city = min(unvisited, key=lambda city: D[current][city])

        route.append(next_city)
        unvisited.remove(next_city)
        current = next_city

    # Return to starting city
    route.append(start)

    # Calculate total distance
    total_distance = route_cost(route, D)

    return route, total_distance


route, distance = nearest_neighbor(D, start=3)

print("Nearest Neighbor Route:", route)
print("Total Distance:", distance)
```

## Comparing Brute Force and Nearest Neighbor

### 1. Results with 4 Cities

We tested both algorithms using the same 4-city distance matrix.

| Algorithm | Distance | Optimal? | Fast for Large Problems? |
|---|---|---|---|
| Brute Force | 80 | Yes | No |
| Nearest Neighbor | 95 | No | Yes |

Brute Force found the shortest route with a distance of **80**, while Nearest Neighbor found a route with a distance of **95**.

This shows that Nearest Neighbor doesn't always find the best route.

### 2. What Happens with More Cities?

When the number of cities increases, Brute Force becomes much slower because it checks every possible route.

For example, with **11 cities**, it needs to check **1,814,400 distinct routes**.

Nearest Neighbor is much faster because it only chooses the closest unvisited city at each step.

### 3. Conclusion

**Brute Force — O(n!):** Always finds the shortest route, but becomes very slow with more cities.

**Nearest Neighbor — O(n²):** Much faster, but doesn't always find the shortest route.

Overall, Brute Force is better for small problems when we need an exact result, while Nearest Neighbor is better when we need a quick solution.
